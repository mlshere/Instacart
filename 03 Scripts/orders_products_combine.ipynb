# 03. Combine products data set

```python
# Import products data
df_prods = pd.read_csv(os.path.join(path, '02 Data', 'Prepared data', 'products_checked.csv'), index_col = False)

df_prods.head()
```

```python
# Drop unnamed column
df_prods.drop(columns='Unnamed: 0', inplace=True)
df_prods.head()
```

```python
# Rename previous _merge as it was causing problems 
df_orders_products_combine.rename(columns={'_merge': 'existing_merge'}, inplace=True)
```

```python
# Merge dataframes
df_ords_prods_merge = df_orders_products_combine.merge(df_prods, on = 'product_id', indicator = True)
```

```python
df_ords_prods_merge['_merge'].value_counts()
```
