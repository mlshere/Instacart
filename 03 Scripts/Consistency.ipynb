````python
# Create a dataframe
df_test = pd.DataFrame()
````

````python
# Create a mixd type column 
df_test['mix'] = ['a', 'b', 1, True]
````

````python
df_test.head()
````

````python
# Check for mixed types

for col in df_test.columns:
    # Get the reference type from the first element of the column
    ref_type = type(df_test[col].iloc[0])
    
    # Check if any element in the column has a type different from the reference type
    weird = df_test[col].apply(type) != ref_type
    
    # If there are any mismatched types, print the column name
    if weird.any():
        print(col)
````

# 04. Missing values

````python
# Find missing values
df_prods.isnull().sum()
````

````python
# Print rows with null values in 'product_name' column
df_nan = df_prods[df_prods['product_name'].isnull() == True]

df_nan
````

````python
# Eliminate null values and create new dataframe
df_prods_clean = df_prods[df_prods['product_name'].isnull() == False]

df_prods_clean.shape
````

# 05. Duplicates

````python
# Finding duplicates

df_dups = df_prods_clean[df_prods_clean.duplicated()]
df_dups
````

````python
# Drop duplicates 
df_prods_clean_no_dups = df_prods_clean.drop_duplicates()

df_prods_clean_no_dups.shape
````

````python
# Exporting data
df_prods_clean_no_dups.to_csv(os.path.join(path, '02 Data', 'Prepared Data', 'products_checked.csv'))
````
