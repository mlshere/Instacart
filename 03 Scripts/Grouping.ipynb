### Target different types of spenders in their marketing campaigns. 

Create a spending flag for each user based on the average price across all their orders using the following criteria:
* If the mean of the prices of products purchased by a user is lower than 10, then flag them as a “Low spender.”
* If the mean of the prices of products purchased by a user is higher than or equal to 10, then flag them as a “High spender.”

```python
# Calculate the mean price per user
user_mean_prices = ords_prods_merge.groupby('user_id')['prices'].mean()
```

```python
# Create column 'mean_price' by mapping user_mean_prices to each user_id

ords_prods_merge['mean_price'] = ords_prods_merge['user_id'].map(user_mean_prices)
```

```python
# Assign spending flags

ords_prods_merge.loc[ords_prods_merge['mean_price'] < 10, 'spending_flag'] = 'Low spender'
ords_prods_merge.loc[ords_prods_merge['mean_price'] >= 10, 'spending_flag'] = 'High spender'
```

```python
# Print unique users
ords_prods_merge[['user_id', 'mean_price', 'spending_flag']].drop_duplicates(subset='user_id').head(15)
```

### Create an order frequency flag that marks the regularity of a user’s ordering behavior

The criteria for the flag should be as follows:
* If the median of “days_since_prior_order” is higher than 20, then the customer should be labeled a “Non-frequent customer.”
* If the median is higher than 10 and lower than or equal to 20, then the customer should be labeled a “Regular customer.”
* If the median is lower than or equal to 10, then the customer should be labeled a “Frequent customer.”

```python
# Calculate the median price per user
user_median_ords = ords_prods_merge.groupby('user_id')['days_since_last_order'].median()
```

```python
# Create column 'median_ords' by mapping user_mean_prices to each user_id
ords_prods_merge['median_ords'] = ords_prods_merge['user_id'].map(user_median_ords)
```

```python
# Assign order frequency flags
ords_prods_merge.loc[ords_prods_merge['median_ords'] <= 10, 'ords_flag'] = 'Frequent customer'
ords_prods_merge.loc[(ords_prods_merge['median_ords'] > 10) & (ords_prods_merge['median_ords'] <= 20), 'ords_flag'] = 'Regular customer'
ords_prods_merge.loc[ords_prods_merge['median_ords'] > 20, 'ords_flag'] = 'Non-frequent customer'
```

```python
ords_prods_merge[['user_id', 'median_ords', 'ords_flag']].drop_duplicates(subset='user_id').head(15)
```
