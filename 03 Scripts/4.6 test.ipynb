# 4.6 Merging and exporting data

### This script contains the following points:

 1. Create a data to experiment on
 2. Concatenate dataframes
 3. Append data
 4. Merge data

```python
# Import libraries
import pandas as pd
import numpy as np
import os
```

### 1. Create data to experiment on 

```python
# Define a dictionary containing January 2020 data 

data1 = {'customer_id':['6732', '767', '890', '635'], 
        'month':['Jan-20', 'Jan-20', 'Jan-20', 'Jan-20'], 
        'purchased_meat':[0, 13, 3, 4], 
        'purchased_alcohol':[1, 2, 10, 0],
        'purchased_snacks': [10, 5, 1, 7]} 
```

```python
# Define a dictionary containing February 2020 data 

data2 = {'customer_id':['6732', '767', '890', '635'], 
        'month':['Feb-20', 'Feb-20', 'Feb-20', 'Feb-20'], 
        'purchased_meat':[0, 10, 5, 3], 
        'purchased_alcohol':[2, 4, 14, 0],
        'purchased_snacks': [15, 3, 2, 6]}
```

```python
# Convert the dictionary into DataFrame
df_1 = pd.DataFrame(data1, index=[0, 1, 2, 3])
df_2 = pd.DataFrame(data2, index=[0, 1, 2, 3])
```

```python
df_1
```

```python
df_2
```

### 2.Concatenate dataframes

```python
# Create a list that contains the dataframes
frames = [df_1, df_2]
```

```python
# Check output
frames
```

```python
# Check data types 
type(frames)
```

```python
# Concatenate the dataframes 

df_concat = pd.concat(frames)
```

```python
# Check output

df_concat
```

```python
# Concatenate the dataframes using axis = 1 (create a wide format as output)

df_concat = pd.concat(frames, axis = 1)

df_concat
```

### 03. Merge data

```python
# Create a new dataframe 
data3 = {'customer_id':['6732', '767', '890', '635'],
    'month':['Jan-20', 'Jan-20', 'Jan-20', 'Jan-20'],
    'days_purchased_on':[0, 10, 4, 1]}

df_3 = pd.DataFrame(data3,index=[0, 1, 2, 3])
```

```python
# Merge df_1 with df_3
df_merged = df_1.merge(df_3, on = 'customer_id')

df_merged
```

```python
# Another merge
df_merged = df_1.merge(df_3, on = ['customer_id', 'month'])

df_merged
```

```python
# Check if costumer_id has same data

df_merged = df_1.merge(df_3, on = ['customer_id', 'month'], indicator = True)

df_merged
```

```python
df_merged['_merge'].value_counts()
```

```python
# Append data
df_appended = df_1.append(df_2)

df_appended
```
