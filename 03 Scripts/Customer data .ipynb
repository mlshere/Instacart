```python
# First I am going to drop the previous '_merge' columns 
ords_prods_merge.drop(columns=['existing_merge', '_merge'], inplace=True)
```

```python
# Merge both dataframes
ords_prods_customer = ords_prods_merge.merge(customer_data, on='user_id', how='inner', indicator=True)
```

```python
ords_prods_customer.shape
```

Due to the size of the dataframe, I have decided that there are some columns that are not needed to have a complete dataframe. I will drop the following columns:
* 'max_order': It was used to classify user loyalty
* 'mean_price': Used to classify user spending habits
* 'median_ords': Used to classify order frequency
* '_merge': Used to get information about the dataframes merge

```python
ords_prods_customer.drop(columns=['max_order', 'mean_price', 'median_ords', '_merge'], inplace=True)
```
